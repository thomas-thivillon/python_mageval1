## 1. Structures de données dans NumPy

`NumPy` (*Numerical Python*) est un package Python couramment utilisé pour les calculs numériques. On l'importe généralement en lui donnat l'alias `np` : 

```python
import numpy as np
```

### 1.1 Les tableaux

NumPy permet de travailler avec une structure de données appelée *tableaux* (`arrays`). Les tableaux ont un usage proche de celui des listes mais leur traitement est plus rapide. On peut utiliser NumPy pour créer des tableaux multidimensionnels (ex: matrices) et effectuer des opérations d'algèbre linéaire sur ces tableaux (ex: multiplication d'une matrice par une autre matrice). 

La création d'un tableau s'effectue avec la méthode `array()`, à partir d'une liste. Dans l'exemple ci-dessous, `x` et `y` sont des tableaux à 1 dimension (des vecteurs) : 

```python
x = np.array([1, 2, 3])

print(x)
print(type(x))
```

On peut aussi créer des matrices (c'est-à-dire des tableaux 2D), avec `np.matrix()` ou `np.array()`, en utilisant des listes imbriquées de même longueur. Utilisons `np.array`. Faites attention à la syntaxe : 

```python
y = np.array([[1,2], [3,4]]) 
print(y)
print(type(y))
```

`x` et `y` possèdent plusieurs _attributs_, accessibles avec `nom.attribut` en remplaçant `nom` par le nom d'objet et `attribut` par le nom de l'attribut qui nous intéresse. Par exemple, on peut vérifier la dimension, le type de données stockées, ou la forme d'un tableau. Pour en savoir plus, consultez `np.array`.

```python
y.ndim # C'est un tableau 2D
```

```python
x.dtype # x est composé d'entiers 64 bits
```

```python
y.shape  # deux lignes et deux colonnes
```

```python
y.size # contient 4 éléments
```

On peut modifier les dimensions d'un tableau en utilisant la fonction `np.reshape()`, sous réserve que les dimensions choisies soient possibles : 

```python
tableau = np.array([1, 2, 3, 4, 5, 6, 7, 8])
np.reshape(tableau, (2,4))
```

Avec des dimensions impossibles : 

```python
np.reshape(tableau, (3,3))
```

### 1.2 Fonctions générant des tableaux

La fonction `np.zeros` crée un tableau 1D dont tous les éléments ont la valeur zéro. On passe le nombre d'éléments `n` souhaité en premier argument et on peut utiliser l'argument optionnel `dtype` pour choisir le type de données souhaité (ex: `ìnt` ou `float`):

```python
print(np.zeros(10))
print(np.zeros(10, dtype = 'int'))
```

Les fonctions `np.empty()` et `np.ones()` produisent des tableaux sur le même principe que `np.zeros()` mais `empty()` crée un tableau dont les valeurs ne sont pas initialisées et `ones()` un tableau dont les valeurs sont initialisées à 1. 

On peut aussi choisir une valeur spécifique avec la fontction `np.full()`:

```python
print(np.full((3,3), 5, dtype = 'int32'))
```

Avec la fonction `np.arange()` on peut générer une séquence de nombres séparés par un interval fixe stockés dans un tableau, puis modifier les dimensions du tableau pour obtenir un matrice. La syntaxe est `np.arange(start, stop, step, dtype)` : 

```python
A = np.reshape(np.arange(2, 33, 2, dtype = 'int'), (4,4))
print(A)
```

### 2. Manipuler les éléments d'un tableau

L'extraction ou la modification des éléments d'un `array` peut se faire grâce à l'indexation comme pour les listes ou les chaines de caractères. 

`A[1,2]` affiche l'élément correspondant à la deuxième ligne et à la troisième colonne de A.

```python
print(A[1,2])
```

La syntaxe `A[1][2]` est équivalente : 

```python
print(A[1][2])
```

Sélectionner plusieurs lignes à la fois : `[1,3]` affiche la deuxième et la quatrième ligne :

```python
A[[1,3]]
```

Pour sélectionner la première et la troisième colonne : `[0,2]` comme second argument entre les crochets. On fournit alors comme premier argument `:`, qui sélectionne toutes les lignes.

```python
A[:,[0,2]]
```

### 2.1 Sélection d'éléments à l'aide de listes

On peut utiliser deux listes pour extraire plusieurs éléments d'un tableau 2D. Ici Python va extraire `A[1,0]` et `A[3,3]`: 

```python
print(A[[1,3],[0,3]])
```

### 2.2 Slicing

On peut sélectionner les lignes d'un tableau à deux dimensions avec la méthode du slicing  : 

```python
A[1:5:2]
```

Pour appliquer la méthode du slicing sur les colonnes : 

```python
print(A[:, 1:5:2])
```

Les deux points avant la virgule indiquent qu'on veut sélectionner toutes les lignes.

### 2.3 Sélections de lignes et de colonnes avec la fonction `np.ix_()`

La syntaxe `A[np.ix_([0,1,3],[0,3])]` permet de filtrer les 1e, 2e et 4e lignes ainsi que les 1e et 4e colonnes de A: 

```python
print(A[np.ix_([0,1,3],[0,3])])
```

### 2.4 Indexation booléenne

Utilisation d'un masque *booléen* pour filtrer les colonnes et les lignes de `A` :

```python
res = A[np.ix_([True, False, True, False], [True, True, True, False])]
print(res)
```

### 3. Modification d'un tableau

Pour remplacer les valeurs des éléments d'un `array`, on utilise le signe `=` : 

```python
B = np.array(A) # B=A crée une référence à A mais pas une copie de A.
B[3,3] = 0 # Donne la valeur 0 à l'élément situé dans la 4e ligne et la 4e colonne
print(B)
```

```python
B[0] = A[3] # Remplace la 1e ligne de B par la 4e de A
print(B)
```

```python
B[:,[3]] = A[:,[3]] # Remplace la 4e colonne de B par la 4e de A
print(B)
```

Pour ajouter des éléments à un objet, on utilise la fonction `np.append()`: 

```python
table_1d = np.array([1,2,3,4])
table_1d = np.append(table_1d, 5)
print(table_1d)
```

Ajout d'une colonne dans une matrice : 

```python
C = np.array(A)
new_col = np.array([[1],[2],[3],[4]])
C= np.append(C, new_col, axis = 1)
print(C)
```

Autre exemple, en utilisant une colonne d'un tableau existant : 

```python
C = np.array(A)
C= np.append(C, B[:,[3]], axis = 1)
print(C)
```

Pour ajouter des lignes, il faut utiliser la fonction `np.vstack()` :

```python
C = np.array(A)
new_line = np.array([34, 36, 38, 40])
C = np.vstack([C, new_line])
print(C)
```

```python
my_line = np.delete(new_line, -1) # Suppression du dernier élément
print(my_line)
```

```python
my_line = np.delete(new_line, [0, 2]) # Suppression des deux premiers éléments
print(my_line)
```

```python
D = np.delete(C, (0), axis=1) # Suppression de la 1ère colonne de C
print(D)
```

```python
D = np.delete(C, (0), axis=0) # Suppression de la 1ère ligne de C
print(D)
```

```python
D = np.delete(C, ([0,2]), axis=0) # Suppression des 2 premières ligne de C
print(D)
```

### 4. Transposition de tableaux

On peut transposer un tableau en utilisant la fonction `np.transpose()` ou l'attribut `T` :

```python
print(D.shape)
```

```python
E = np.transpose(D)
print(E.shape)
print(E)
```

```python
F = D.T
print(F.shape)
print(F)
```

Attention, les deux méthodes créent des références à l'objet d'origine (elles le modifient) quand on assigne leur résultat à un nouvel objet : 

```python
D[0,0] = 0
print(D)
print(E)
print(F)
```

### 5. Opérations sur les tableaux

On peut utiliser les opérateurs `+`, `-`, `*` et `\` pour effectuer des opérations élément par élément entre deux tableaux : 

```python
tab_1 = np.array(range(1,5))
tab_1 = np.reshape(tab_1, (2,2))

tab_2 = np.full((2,2), 2)

tab_3 = tab_1 + tab_2

tab_4 = tab_1 / tab_2

print("tab_1: \n",tab_1)
print("tab_2: \n",tab_2)
print("tab_3: \n",tab_3)
print("tab_4: \n",tab_4)
```

On peut également effectuer des opérations entre un tableau et un scalaire. Dans ce cas le scalaire est ajouté/soustrait à chaque élément du tableau ou chaque élément est multiplié/divisé par le scalaire: 

```python
tab_5 = tab_1 + 3
print("tab_5: \n",tab_5)

tab_6 = tab_1 * 3
print("tab_6: \n",tab_6)
```

Pour effectuer un produit de matrices, on utilise la fonction `np.dot()`. Il faut au préalable vérifier que les matrices ont des dimensions compatibles : 

```python
print(np.dot(tab_2, tab_1))
```

### 1.9 Génération de variables pseudo-aléatoires

Nous allons générer des données aléatoires pendant ce cours. Pour cela, nous utiliserons principalement la fonction `np.random.normal()` pour créer des vecteurs de variables aléatoires normales. La fonction prend 3 arguments : `normal(loc=0.0, scale=1.0, size=None)`. Par défaut, elle génère des variables aléatoires normales de moyenne (`loc`) **0**, d'écart-type (`scale`) **1**, et une seule variable sauf si l'on modifie `size`.

Nous générons maintenant 100 variables aléatoires indépendantes suivant une loi ***N*****(0,1)**.

```python
# Exécutez-la plusieurs fois, que remarquez-vous ?
x = np.random.normal(size=100)
x
```

Pour que notre code donne exactement les mêmes résultats à chaque exécution, on peut fixer une graine aléatoire (seed) avec la fonction `np.random.default_rng()`. Si l'on fixe une graine avant de générer des données aléatoires, réexécuter le code donnera les mêmes résultats. Ainsi, pour générer des données normales, on utilise `rng.normal()`.

```python
rng = np.random.default_rng(1303)
print(rng.normal(scale=5, size=2))
rng2 = np.random.default_rng(1303)
print(rng2.normal(scale=5, size=2))
```
