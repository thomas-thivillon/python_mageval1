## Partie 2 (/10 points)

Ecrivez un programme qui utilise une boucle pour vérifier si les entiers `[7, 12, 15, 18, 20, 21, 25, 36, 42]` sont divisibles par 2 et par 3, par un seul des deux, ou par aucun des deux. Pour un entier `k` donné, le programme doit afficher :
- "`k` est divisible par 2 et par 3" si l'entier est divisible par 2 et par 3 (ex: "12 est divisible par 2 et par 3").
- "`k` est divisible par 2 ou par 3 mais pas par les deux" si l'entier est divisible par 2 mais pas par 3 ou par 3 mais pas par 2.
- "`k` n'est divisible ni par 2 ni par 3" si aucune des deux affirmations précédentes n'est vraie.

Après exécution du programme, la sortie de Python doit contenir 9 phrases correspondant à l'évaluation de chacun des 9 entiers.

Aide : Vous pouvez utiliser l'opérateur `%` qui permet de calculer le reste de la division euclidienne d'une variable par un nombre. Ex: `k%2 == 0` est une expression booléenne qui est `True` si le reste de la division de `k` par $2$ est $0$.
