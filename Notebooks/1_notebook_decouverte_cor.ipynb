## 3. Gérer les cellules : créer, supprimer, déplacer

En utilisant uniquement les raccourcis clavier (mode commande) :

* créez deux nouvelles cellules de code en dessous de la cellule Markdown précédente ;
* supprimez l'une des deux cellules que vous venez de créer ;
* déplacez la cellule restante pour qu'elle se retrouve juste en dessous de votre toute première cellule de code (celle du point 1).

**Raccourcis clavier :**

- nouvelle cellule de code en-dessous d'une cellule active : `alt` + `entrée`
- supprimer une cellule active : `esc` + 2 x `d`
- déplacer une cellule active vers le haut ou vers le bas : `ctrl` + `shift` + `flèche du haut` ou `flèche du bas`
