## 1.2 Variables

### 1.2.1 Utiliser des variables

Lorsque l'on veut que Python stocke un résultat en mémoire, il faut l'assigner à une *variable*. Pour créer une variable, on utilise le signe égal (=) et on assigne le résultat (ou tout autre objet) au nom de la variable selon la syntaxe suivante : `nom_variable = objet`.

On peut afficher le contenu d'une variable à l'aide de la fonction `print()` :

`print(nom_variable)`

Ou tout simplement en tapant le nom de la variable :

```python
x = 7
print(x)
x
```

Modifiez la valeur de la variable avec une nouvelle assignation :

```python
x = 8
x
```

```python
x = "Madagascar"
x
```

Il est possible de créer des copies d'une variable. La variable originale ne sera pas affectée par les modifications apportées à la copie :

```python
y = x
y
```

```python
y = "Antananarivo"
print(y)
print(x)
```

Une variable peut être supprimée avec l'instruction `del` :

```python
del y 
y
```

Les variables peuvent être incrémentées à l'aide des opérateurs spéciaux `+=` et `*=` :

```python
x = 0
x += 1
print(x)
x += 1
print(x)
```

```python
y = 1
y *= 3
print(y)
y *= 3
print(y)
```

### 1.2.2 Conventions de nommage des variables

Comme dans la plupart des logiciels de data science et de statistiques :

- Les noms de variables ne peuvent être composés que de caractères alphanumériques et du tiret bas (_).
- Les noms de variables ne peuvent pas commencer par un caractère numérique.
- Il est interdit d'inclure des espaces dans les noms de variables.

De plus, pour améliorer la lisibilité, il est recommandé de n'utiliser que des lettres minuscules et de séparer les mots par un tiret bas dans les noms de variables (c'est-à-dire utiliser `nom_variable` et non `nomvariable` ou `Nom_Variable`).

## 2.2 Chaînes de caractères

Une chaîne de caractères (`string`) est une suite de caractères [Unicode](https://en.wikipedia.org/wiki/Unicode) tels que des lettres, des chiffres, des signes de ponctuation, etc.

Les chaînes de caractères sont définies à l'aide de guillemets simples ou doubles :

```python
#Ceci fonctionne aussi :
y = 'it works'

#Ceci ne fonctionne pas :
z = 'I'm a genius'

print(y)
print(z)
```

Les chaînes de caractères peuvent être concaténées...

```python
"This is" + " " + "fun"
```

... et répétées :

```python
"fun " * 3 + "... !"
```

Chaque caractère d'une chaîne possède un index correspondant à sa position dans la séquence. On peut utiliser cet index pour extraire des caractères spécifiques d'une chaîne. **Important : l'indexation commence à 0 en Python**

```python
#Extract 1st element of the sequence
"This is still fun !"[0]
```

```python
#Extract 2nd element of the sequence
"This is still fun !"[1]
```

```python
#Extract last element of the sequence
"This is still fun !"[-1]
```

```python
#Extract everything from a certain character
"This is still fun !"[8:]
```

```python
#Extract everything up to a certain character
"This is still fun !"[:7]
```

```python
#Extract a substring
"This is still fun !"[14:17]
```

```python
#Extract every other element of the sequence
"This is still fun !"[0::2]
```

```python
#Extract every three element of the sequence starting from the second and finishing at the 17th
"This is still fun !"[1:17:2]
```

```python
#Reverse a sequence
"This is still fun !"[::-1]
```

La syntaxe générale est `x[a:b:c]` où `a` est la position du caractère de départ, `b` la position du caractère de fin, et `c` le pas d'indexation (rappel : **l'indexation commence à 0**).

# 3. Structures

Les principaux types de structures en `Python` sont :

- les listes,
- les tuples,
- les sets (ensembles),
- et les dictionnaires.

On peut voir les structures de données comme des conteneurs, car elles permettent de stocker, organiser et accéder à des données.

Les listes et les tuples sont des **conteneurs séquentiels** : les éléments qu'ils contiennent sont **ordonnés**, et leur position est enregistrée dans un **index**.

Les dictionnaires et les sets sont des **structures de données non ordonnées** : les objets ne sont plus stockés par position (ou index) mais par **clé**, qui est un identifiant unique.

## 3.1 Listes

Les `listes` sont l'une des structures les plus flexibles de Python. Une `liste` est un regroupement de valeurs.
La création d'une liste se fait en écrivant les valeurs séparées par une virgule
et encadrées par des crochets ([ et ]). Vous pouvez stocker aussi bien des valeurs numériques que des chaînes de caractères dans des listes.

Demandez à Python d'assembler les nombres 3, 4 et 5, et de les enregistrer sous forme de liste nommée `x`. Affichez ensuite le contenu de `x` :

```python
x = [3, 4, 5]
x
```

Copier une liste : **attention**, copier une liste en Python se fait à l'aide de la fonction `list()` et non en utilisant le signe égal :

```python
y=x
y[0] = 0
print(y)
print(x)
```

```python
x = [3, 4, 5]
y = list(x)
y[0] = 0
print("x : ", x)
print("y : ", y)
```

## 3.2 Tuples

Les `tuples` sont des séquences d'objets Python. Pour créer un tuple, on énumère les valeurs séparées par des virgules. Contrairement aux listes, **les tuples sont immuables** (c'est-à-dire qu'ils ne peuvent pas être modifiés après avoir été créés).

```python
x = 1, 4, 9, 16, 25
print(x)
```

```python
x[0] = 1
```

## 3.3 Sets

Les `sets` sont des collections non ordonnées d'éléments uniques. Les sets sont immuables. Contrairement aux tuples, les sets ne sont pas
indexés :

```python
new_set = {"Bordeaux", "Aix-en-Provence", "Nice", "Rennes"}
print(new_set)
```

Les sets sont assez peu utilisés en pratique, mais ils sont très utiles dans
certaines situations spécifiques. Grâce à l'unicité des éléments qu'ils
contiennent, les sets permettent de supprimer simplement et efficacement les doublons d'
un conteneur séquentiel, comme une liste.

**Déduplication**

Supposons que vous vouliez supprimer les doublons d'une liste donnée. Par définition,
convertir une liste en set supprime les doublons. Cependant, on souhaite généralement
revenir à une liste, car les sets n'offrent pas la même flexibilité que les
listes (par exemple, la possibilité de retrouver un élément par sa position). Il est
donc courant d'effectuer l'enchaînement d'opérations `liste -> set -> liste`
pour dédupliquer une liste.

```python
l = [1, 2, 3, 3, 2, 1]
l_dedup = list(set(l))
l_dedup
```

**Opérations sur les sets**

Puisque les sets représentent informatiquement des ensembles mathématiques, il n'est
pas surprenant qu'ils permettent d'effectuer des opérations ensemblistes élémentaires. Par exemple,
l'union et l'intersection.

```python
l1 = [5, 3, 2, 3, 3, 5, 8, 9]
l2 = [3, 7, 0, 0, 1, 9, 4, 6]

# Union: elements in l1, l2, or both
l_union = list(set(l1) | set(l2))
print(l_union)

# Intersection: elements in both l1 and l2
l_inter = list(set(l1) & set(l2))
print(l_inter)
```

**Tests d'appartenance**

Les sets sont également très utiles pour les tests d'appartenance, car ils offrent de
bien meilleures performances que les listes pour ce type de test.

La notion de test sera abordée dans un prochain tutoriel. Pour l'instant,
notons qu'un test d'appartenance tel que « l'élément `a` est-il dans la liste `l` »
s'écrit en Python `a in l` et renvoie `True` ou `False` selon
que `a` est effectivement présent ou non dans la liste `l`.

```python
l = [1, 2, 3]
2 in l
```

```python
4 in l
```

## 3.4 Dictionnaires

Les `dictionnaires` Python sont une implémentation des `objets clé-valeur`, les clés étant
indexées. Les clés sont souvent du texte, les valeurs peuvent être de différents types et structures. Pour créer un dictionnaire, on peut utiliser des accolades ({}) :

```python
my_dict = { "nom": "Kyrie",
"prenom": "John",
"naissance": 1992,
"equipes": ["Cleveland", "Boston"]}
print(my_dict)
```

```python
print(my_dict["equipes"])
```

```python
my_dict["equipes"] = ["Montclair Kimberley Academy",
"Cleveland Cavaliers", "Boston Celtics"]
print(my_dict)
```

Les méthodes `.keys()`, `.values()` et `.items()` renvoient respectivement les clés,
les valeurs et les paires clé-valeur d'un dictionnaire. Les objets
renvoyés par ces méthodes sont un peu complexes, mais ils peuvent être convertis
en listes à l'aide de la fonction `list` pour les interroger par position :

```python
list(my_dict.keys())
```

```python
list(my_dict.values())
```

```python
list(my_dict.items())
```
