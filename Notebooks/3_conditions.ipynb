# Séance 3 : Conditions et structures conditionnelles

## 1. Conditions

Nous allons maintenant détailler le fonctionnement des conditions en Python.
Elles constituent une étape majeure dans la création de programmes qui automatisent des opérations,
car elles permettent d'**exécuter ou non du code selon certaines
conditions**. Elles permettent à l'ordinateur de prendre des décisions selon des
critères fixés par l'utilisateur.

### 1.1 Le type booléen

Dans sa forme la plus simple, un `test` en Python est une expression qui s'évalue
à une valeur `booléenne` (voir ci-dessus).

### 1.2 Opérateurs de comparaison

Les opérateurs de comparaison formalisent les opérations mathématiques de comparaison
(égalité, inégalité, inéquations). Ils **comparent deux valeurs et
renvoient une valeur booléenne**.

Opérateur \| Signification \|  
`|` \|  
== \| Égal à \|  
!= \| Différent de \|  
\< \| Inférieur à \|  
\> \| Supérieur à \|  
\<= \| Inférieur ou égal à \|  
\>= \| Supérieur ou égal à \|

Illustrons ces opérateurs avec quelques exemples.

```python
1 == 1
```

```python
1 == 2
```

```python
1 != 2
```

Tout semble fonctionner correctement pour les opérations mathématiques usuelles.
Mais ces opérateurs fonctionnent en réalité sur tout type d'objet.

```python
'do re mi fa sol' == 'do re mi fa sol'
```

```python
'do re mi fa sol' == 'Do Re Mi Fa Sol'
```

```python
[1, 2, 3] == [1, 2, 3]
```

Enfin, il est possible de chaîner des comparaisons. L'expression renvoie
`True` à condition que chacune des comparaisons soit vraie.

```python
1 < 2 <= 3
```

```python
4 < 5 <= 3
```

### 1.3 Opérateurs booléens

Les opérateurs booléens permettent de **tester simultanément plusieurs expressions
logiques**. Fondamentalement, ces opérateurs prennent deux valeurs booléennes en
entrée et renvoient une seule valeur booléenne selon des règles logiques fixes.
Ces règles sont énoncées dans des **tables de vérité**.

### Opérateur `and`

Le premier opérateur booléen est `and`. Regardons sa table de vérité :

| Expression  | Évaluation |
|-------------|------------|
|`True and True` | `True`|
| `True and False` |  `False` |
| `False and True` |  `False` |
| `False and False` |  `False`  |

Vérifions ces règles en pratique avec quelques exemples.

```python
True and True
```

```python
True and False
```

```python
(2 > 1) and (3 <= 4)
```

```python
a = ("x" != "z")
b = ("x" == "y")
a and b
```

Remarquez l'utilisation des parenthèses pour délimiter les tests : elles ne sont pas obligatoires
mais sont fortement recommandées car elles améliorent grandement la lisibilité des
tests.

### Opérateur `or`

Le deuxième opérateur booléen est `or`. Sa table de vérité est la suivante :


| Expression  | Évaluation |
|-------------|------------|
|`True or True` | `True`|
| `True or False` |  `True` |
| `False or True` |  `True` |
| `False or False` |  `False`|

```python
a = ("x" != "z")
b = ("x" == "y")
a or b
```

### Opérateur `not`

Le dernier opérateur booléen est `not`. Sa table de vérité est la suivante :

| Expression  | Évaluation |
|-------------|------------|
| `not True`  | `False`    |
| `not False` | `True`     |

```python
not (3 + 3 == 6)
```

```python
not (1 < 1)
```

## 2 Structures conditionnelles

Toutes les expressions que nous avons vues jusqu'à présent sont des expressions booléennes : un test
est effectué, et l'opération renvoie `True` ou `False` selon
que l'expression évaluée est vraie ou non. Dans le contexte d'un
programme informatique qui automatise des opérations, on souhaite les utiliser comme
des **conditions** : *si* l'expression est vraie, alors l'ordinateur doit
effectuer une certaine opération. Les **structures conditionnelles** permettent précisément
cet usage.

Illustrons ce principe en implémentant le programme suivant :

-   Définissez une variable `x`.

-   Si `x` est supérieur à `une valeur de votre choix`, affichez « L'expression est vraie. »

-   Sinon, affichez « L'expression est fausse. »

Faites varier la valeur de `x` pour vérifier le bon fonctionnement du test.

```python
x = 12

if x >= 12:
    print("L'expression est vraie.")
else:
    print("L'expression est fausse.")
```

### 2.1 Blocs d'instructions et indentation

L'exemple précédent illustre la syntaxe des structures conditionnelles en
Python. Ces structures reposent sur des *blocs d'instructions*, qui
délimitent l'ensemble des instructions à exécuter lorsqu'un test est vrai.
Les structures conditionnelles obéissent à trois règles :

-   La ligne spécifiant le test se termine par `:`.

-   Toutes les instructions devant être exécutées si le test est vrai se trouvent au
    même niveau d'indentation.

-   La structure conditionnelle se termine lorsque l'indentation revient à son
    niveau initial.

Notez que les structures conditionnelles peuvent effectivement être imbriquées, comme l'illustre
l'exemple suivant.

```python
x = 7

if x >= 5:
    print("La première expression est vraie.")
    if x >= 6:
        print("La seconde expression est vraie.")
```

### 2.2 Les instructions `if`, `else` et `elif`

Dans les structures conditionnelles, les tests peuvent être spécifiés à l'aide de trois
instructions : `if`, `else` et `elif`. Les exemples précédents ont déjà
illustré le fonctionnement de la première (et la plus fréquente)
instruction.

En pratique, on souhaite souvent spécifier une alternative lorsque la condition
de l'instruction `if` renvoie `False`. L'instruction `else` permet de
spécifier un bloc d'instructions alternatif.

```python
city = "Antananarivo"

if city in ["Antananarivo", "Fianarantsoa", "Tulear"]:
    print("Ville connue.")
else:
    print("Ville inconnue.")
```

Enfin, on peut vouloir spécifier plusieurs alternatives. Dans ce cas, on
utilisera des instructions `elif`. La première instruction `elif` ne s'exécutera
que si le test de l'instruction `if` renvoie `False`. La seconde instruction
`elif` ne s'exécutera que si le test de la première instruction `elif`
renvoie `False`, et ainsi de suite. Là encore, une instruction finale `else`
peut être spécifiée ; elle ne s'exécute que si aucun des tests précédents n'a renvoyé
`True`.

```python
city = "Antananarivo"

if city == "Tulear":
    print("Je vais à Tuléar.")
elif city == "Fianarantsoa":
    print("Je vais à Fianarantsoa.")
elif city == "Antananarivo":
    print("Je vais à Antananarivo.")
else:
    print("Je ne sais pas où je vais.")
```
