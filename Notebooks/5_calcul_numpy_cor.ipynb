# Exercices notebook 5 : vecteurs et matrices avec NumPy — Corrigé

On utilise les tableaux (`array`) du module NumPy, importé sous l'alias habituel `np`.

```python
import numpy as np
```

## Exercice 1 : manipulation d'un vecteur

**1.** Créer le vecteur `x = [1, 2, 3, 4, 5]` à l'aide d'un tableau.

```python
x = np.array([1, 2, 3, 4, 5])
x
```

**2.** Type de `x`, puis sa longueur.

```python
print(type(x))
print(len(x))      # équivalent ici à x.size ou x.shape[0]
```

**3.** Premier élément, puis dernier élément. En Python, les positions commencent à 0 ; l'indice `-1` désigne le dernier élément.

```python
print(x[0])
print(x[-1])       # ou x[len(x) - 1]
```

**4.** Les trois premiers éléments, stockés dans `a`. Dans `x[:3]`, la borne de fin (position 3) est exclue.

```python
a = x[:3]
a
```

**5.** Les 1er, 2e et 5e éléments, c'est-à-dire les **positions** 0, 1 et 4, stockés dans `b`. On passe une liste de positions entre crochets.

```python
b = x[[0, 1, 4]]
b
```

**6.** Ajouter 10 à `x`, puis multiplier le résultat par 2. L'opération est appliquée à chaque élément.

```python
(x + 10) * 2
```

**7.** Addition de `a` et `b`.

```python
a + b
```

Les deux vecteurs ont la même longueur (3) : l'addition se fait **élément par élément** : `[1+1, 2+2, 3+5]`.

**8.** Addition `x + a`, puis `a + x`. Les deux vecteurs n'ont pas la même longueur (5 et 3) : NumPy ne sait pas les apparier et renvoie une erreur. On l'intercepte avec `try` / `except` pour pouvoir afficher le message sans interrompre le notebook.

```python
try:
    x + a
except ValueError as e:
    print("x + a :", e)

try:
    a + x
except ValueError as e:
    print("a + x :", e)
```

Le résultat est le même dans les deux sens : une erreur *operands could not be broadcast together*. Contrairement à R, NumPy ne « recycle » pas le vecteur le plus court.

**9.** Multiplier le vecteur par le scalaire `c = 2`.

```python
c = 2
x * c
```

**10.** Multiplication de `a` et `b`.

```python
a * b
```

Comme pour l'addition, `*` multiplie **élément par élément** : `[1×1, 2×2, 3×5]`. Ce n'est pas un produit scalaire ; celui-ci s'obtient avec `a @ b` ou `np.dot(a, b)`.

```python
np.dot(a, b)
```

**11.** Multiplication `x * a`.

```python
try:
    x * a
except ValueError as e:
    print("x * a :", e)
```

Même problème qu'à la question 8 : les longueurs (5 et 3) sont incompatibles, d'où une erreur.

**12.** Positions des multiples de 2, stockées dans `ind`, puis multiples de 2 de `x`, stockés dans `mult_2`. L'opérateur `%` donne le reste de la division entière : `x % 2 == 0` est un vecteur de booléens.

```python
x % 2 == 0
```

```python
ind = np.where(x % 2 == 0)[0]   # np.where renvoie un tuple, on prend son premier élément
ind
```

```python
mult_2 = x[ind]          # ou directement : x[x % 2 == 0]
mult_2
```

**13.** Éléments de `x` multiples de 3 **et** de 2. On combine les conditions avec `&`, en mettant chaque condition entre parenthèses.

```python
x[(x % 3 == 0) & (x % 2 == 0)]
```

Aucun élément de `x` n'est multiple de 6 : on obtient un tableau vide.

**14.** Éléments de `x` multiples de 3 **ou** de 2 : opérateur `|`.

```python
x[(x % 3 == 0) | (x % 2 == 0)]
```

**15.** Somme des éléments de `x`.

```python
x.sum()            # ou np.sum(x)
```

**16.** Remplacer le premier élément de `x` par 4.

```python
x[0] = 4
x
```

Remarque : `a` a été créé par découpage (`x[:3]`). C'est une **vue** sur `x` et non une copie : il est donc modifié lui aussi. Ce n'est pas le cas de `b`, créé à partir d'une liste de positions. Pour obtenir une copie indépendante, on écrit `x[:3].copy()`.

```python
print(a)
print(b)
```

**17.** Remplacer le premier élément de `x` par `NaN`, puis calculer la somme.

`NaN` (*Not a Number*) est un nombre décimal (`float`). Or `x` contient des entiers : `x[0] = np.nan` provoquerait une erreur. Il faut d'abord convertir `x` en décimaux avec `astype(float)`.

```python
x = x.astype(float)
x[0] = np.nan
x
```

```python
print(x.sum())         # NaN : la valeur manquante « contamine » la somme
print(np.nansum(x))    # somme en ignorant les NaN
```

**18.** Supprimer le vecteur `x`.

```python
del x
```

Après `del x`, le nom `x` n'existe plus : `print(x)` provoquerait une erreur `NameError`.

```python
try:
    print(x)
except NameError as e:
    print(e)
```

## Exercice 2 : manipulation d'une matrice

**1.** Créer la matrice $A$. On donne la liste de ses lignes.

```python
A = np.array([[-3,  5,  6],
              [-1,  2,  2],
              [ 1, -1, -1]])
A
```

**2.** Dimension, nombre de colonnes, nombre de lignes et longueur.

```python
print(A.shape)       # (lignes, colonnes)
print(A.shape[1])    # nombre de colonnes
print(A.shape[0])    # nombre de lignes
print(len(A))        # longueur : nombre de lignes pour un tableau 2D
print(A.size)        # nombre total d'éléments
```

Attention : pour une matrice, `len()` donne le nombre de **lignes** (3), alors que `length()` en R donne le nombre total d'éléments (9). L'équivalent NumPy est `A.size`.

**3.** Deuxième colonne de $A$ (position 1), puis première ligne (position 0). Le symbole `:` signifie « toutes les lignes » ou « toutes les colonnes ».

```python
A[:, 1]
```

```python
A[0, :]          # ou simplement A[0]
```

**4.** Élément en troisième position de la première ligne.

```python
A[0, 2]
```

**5.** Sous-matrice $2 \times 2$ du coin inférieur droit : lignes et colonnes à partir de la position 1.

```python
A[1:, 1:]
```

**6.** Somme des colonnes (`axis=0` : on somme le long des lignes), puis somme des lignes (`axis=1`).

```python
print(A.sum(axis=0))   # une somme par colonne
print(A.sum(axis=1))   # une somme par ligne
```

**7.** Diagonale de $A$.

```python
np.diag(A)       # ou A.diagonal()
```

**8.** Ajouter le vecteur colonne $(1, 2, 3)^\top$ à droite de $A$ et stocker le résultat dans `B`. Le vecteur doit être mis sous forme de colonne (3 lignes, 1 colonne) avec `reshape(-1, 1)`, puis on concatène horizontalement.

```python
v = np.array([1,2,3])
B = np.append(A, v.reshape(-1,1), axis = 1)  # ou np.column_stack((A, v))
B
```

**9.** Retirer le quatrième vecteur (colonne en position 3) de `B`. `np.delete` renvoie une copie : on réaffecte le résultat à `B`.

```python
B = np.delete(B, (3), axis=1)
B
```

**10.** Retirer la première et la troisième ligne de `B` (positions 0 et 2).

```python
B = np.delete(B, ([0, 2]), axis=0)
B
```

**11.** Ajouter le scalaire 10 à $A$ : 10 est ajouté à chaque élément.

```python
A + 10
A
```

**12.** Ajouter le vecteur colonne $(1, 2, 3)^\top$ à $A$. Sous forme de colonne, le vecteur est ajouté à chaque colonne de $A$ : on ajoute 1 à la première ligne, 2 à la deuxième et 3 à la troisième.

```python
A + v.reshape(-1,1)
```

Attention : sans `reshape`, `v` est traité comme une **ligne** et ajouté à chaque ligne de $A$ (1 à la première colonne, 2 à la deuxième, 3 à la troisième). Le résultat est différent :

```python
A + v
```

**13.** Ajouter la matrice identité $I_3$ à $A$.

```python
A + np.eye(3)        # np.eye(3) crée I_3 (avec des décimaux)
```

**14.** Diviser tous les éléments de $A$ par 2.

```python
A / 2
```

**15.** Multiplier $A$ par le vecteur $(1, 2, 3)^\top$. Il s'agit du produit matriciel $A v$, obtenu avec l'opérateur `@` (ou `np.dot`). Le résultat est un vecteur de longueur 3.

```python
A @ v                # ou np.dot(A, v)
```

À ne pas confondre avec `A * v`, qui multiplie élément par élément (chaque ligne de $A$ par `v`) :

```python
A * v
```

**16.** Transposée de $A$.

```python
A.T                  # ou np.transpose(A)
```

**17.** Produit $A^\top A$.

```python
A.T @ A
```
