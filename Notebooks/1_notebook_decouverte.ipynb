# Exercice : Prise en main d'un notebook Jupyter avec Python

Cet exercice vous permettra de vous familiariser avec les manipulations de base d'un notebook Jupyter : gestion des cellules, raccourcis clavier, syntaxe Markdown et quelques manipulations Python simples.

---

## 1. Créer et exécuter une première cellule de code

Sous ce texte, créez une cellule de code. Écrivez un code Python qui affiche la phrase :

```
Bienvenue dans un notebook !
```

Exécutez la cellule (`Shift + Entrée`). Modifiez ensuite votre code pour afficher une autre phrase, puis ré-exécutez la cellule.


## 2. Passer d'une cellule de code à une cellule Markdown (raccourci clavier)

Créez une nouvelle cellule. Sans utiliser la souris ni le menu déroulant, transformez-la en cellule Markdown à l'aide d'un raccourci clavier (vous pouvez faire une recherche Google pour trouver le raccourci clavier).

Dans cette cellule Markdown, écrivez un court texte comportant :

* un titre de niveau 2 (équivalent du `<h2>` en HTML) ;
* un sous-titre de niveau 3 ;
* un mot en gras et un mot en italique ;
* une liste à puces (non ordonnée) avec au moins 3 éléments ;
* une petite équation, par exemple `y = ax + b` ou une équation de votre choix (en cherchant sur internet comment écrire des équations en Markdown/LaTeX si besoin). Attention, l'exemple d'équation que je vous ai donné n'est volontairement pas au format "équation" (math mode). 

Exécutez la cellule pour voir le résultat mis en forme.

## 3. Gérer les cellules : créer, supprimer, déplacer

En utilisant uniquement les raccourcis clavier (mode commande) :

* créez deux nouvelles cellules de code en dessous de la cellule Markdown précédente ;
* supprimez l'une des deux cellules que vous venez de créer ;
* déplacez la cellule restante pour qu'elle se retrouve juste en dessous de votre toute première cellule de code (celle du point 1).

## 4. Stocker et afficher des chaînes de caractères (`string`)

Créez une cellule de code. Stockez la phrase suivante dans un objet nommé `phrase` :

```
J'apprends Python
```

Affichez le contenu de `phrase`.

*Remarque : cette phrase contient une apostrophe. Réfléchissez à la manière d'écrire cette chaîne de caractères sans provoquer d'erreur.


## 5. Additionner deux entiers et afficher le résultat

Créez une nouvelle cellule de code. Stockez deux nombres entiers de votre choix dans deux objets, par exemple `a` et `b`. Calculez leur somme et stockez le résultat dans un troisième objet, par exemple `resultat`.

Affichez ensuite la valeur de `resultat` de deux manières différentes :

1. en utilisant la fonction `print()` ;
2. en écrivant simplement le nom de l'objet en dernière ligne de la cellule (affichage automatique du notebook).

## 6. Renommer et exporter

Renommez votre notebook "notebook_[votre_nom]" et téléchargez-le.
