### 1. Convertir une boucle `for` en boucle `while`

Réécrivez la boucle `for` suivante à l'aide d'une boucle `while`.

```python
gamme = ['do', 're', 'mi', 'fa', 'sol', 'la', 'si']

for i, note in enumerate(gamme):
    print("La note numéro " + str(i) + " de la gamme de do majeur est " + note)
```

```python
i = 0
while i <= (len(gamme) - 1): 
    note = gamme[i]
    print("La note numéro " + str(i) + " de la gamme de do majeur est " + note)
    i += 1
```

### 2. Trouver un élément dans une liste

Étant donné un entier cible `target_num` et une liste d'entiers `l` telle que
définie dans la cellule suivante. À l'aide d'une boucle `for` et de la fonction
`enumerate` :

-   Vérifiez si l'entier cible est présent dans la liste `l`.

-   Si oui, affichez le message « Le nombre `target_num` est à la position
    `i` dans la liste », et terminez la boucle.

```python
target_num = 78

l = [12, 98, 65, 39, 78, 55, 119, 27, 33]

for idx, num in enumerate(l):
    if target_num == num:
        i= idx + 1
        print("Le nombre ", target_num, " est en position ", i, " dans la liste.")
        break
```
